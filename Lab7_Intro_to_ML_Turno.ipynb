## Description
This laboratory compares superrvised and unsupervised learning, demonstrates overfitting and underfitting through a complexity sweep, and evaluates models using k-fold cross-validation.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    adjusted_rand_score,
)
```

```python
GITHUB_RAW_URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(GITHUB_RAW_URL)

print("Dataset shape:", df.shape)
print()
print(df.head())
print()
print("Class distribution:")
print(df["species"].value_counts())
```

```python
feature_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
]
X = df[feature_columns].values
y = df["species"].values

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    stratify = y,
    random_state=42
)

print("Training sample:", X_train.shape)
print("Test sample:", X_test.shape)
```

## Part A: Supervised Learning (Labels ARE Used)

The model learns a mapping from features to known labels.
Scaling is placed INSIDE the pipeline so that each cross-validation fold computes scaling statistics from its training portion only.
This prevent data leakage.

```python
supervised_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", DecisionTreeClassifier(max_depth = 3, random_state = 42))
])

supervised_pipeline.fit(X_train, y_train)
superised_predictions = supervised_pipeline.predict(X_test)

print("SUPERISED LEARNING")
print("=" * 50)
print("Test accuracy:", accuracy_score(y_test, superised_predictions))
print()
print("Confusion Matrix:")
print(confusion_matrix(y_test, superised_predictions))
print()
print(classification_report(y_test, superised_predictions))
```

## Part B: Unsupervised Learning (Labels ARE NOT Used)

The species column is completely ignored. K-Means must discover group structure on its own.
Afterwards, we compare the discovered clusters to the true labels only for evaluation purposes.

```python
#NOTE: y is NOT used anywhere in this part.

X_unlabeled = df[feature_columns].values

kmeans = KMeans(n_clusters = 3, n_init = 10, random_state = 42)
cluster_labels = kmeans.fit_predict(X_unlabeled)

comparison = pd.crosstab(
    pd.Series(cluster_labels, name = "Discovered Clusters"),
    pd.Series(y, name = "True Species")
)
print("UNSUPERVISED LEARNING")
print("=" * 50)
print("Cluster vs. Species Comparison:")
print(comparison)
print()
print("Adjusted Rand Index (cluster agreement with true labels):")
print(adjusted_rand_score(y, cluster_labels))
```

```python
pca = PCA(n_components = 2)
X_pca = pca.fit_transform(X_unlabeled)

plt.figure(figsize = (12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X_pca[:, 0], X_pca[:, 1], c = cluster_labels, cmap="viridis", edgecolors="k")
plt.title("K-Means Clusters (unsupervised)")
plt.xlabel("PC1")
plt.ylabel("PC2")

plt.subplot(1, 2, 2)
species_codes = pd.factorize(y)[0]
plt.scatter(X_pca[:, 0], X_pca[:, 1], c = species_codes, cmap="viridis", edgecolors="k")
plt.title("True Species (for comparison only)")
plt.xlabel("PC1")


plt.tight_layout()
plt.show()

print("Explained variance ratio:", pca.explained_variance_ratio_)
```

## Part  C: Complexity Sweep (Bias - Variance Demonstration)

We train decision trees with increaseing max_depth.

- Very small depth -> high bias -> underfitting

- Moderate depth -> sweet spot -> good generalization

- Very Large depth -> high variance -> overfitting (train = 100%, gap widens)

```python
depth_values = list(range(1, 16))

train_scores =[]
test_scores = []

for depth in depth_values:
  pipe =Pipeline([
      ("scale", StandardScaler()),
      ("model", DecisionTreeClassifier(max_depth = depth, random_state = 42))
  ])
  pipe.fit(X_train, y_train)
  train_scores.append(accuracy_score(y_train, pipe.predict(X_train)))
  test_scores.append(accuracy_score(y_test, pipe.predict(X_test)))

sweep_table = pd.DataFrame({
    "Max Depth": depth_values,
    "Training Accuracy": train_scores,
    "Test Accuracy": test_scores,
    "gap": np.array(train_scores) - np.array(test_scores)
})

print(sweep_table.to_string(index = False))

```

```python
plt.figure(figsize=(9, 5))

plt.plot(depth_values, train_scores, marker="o", label="Training accuracy")
plt.plot(depth_values, test_scores, marker="s", label="Testing accuracy")

plt.axvspan(1, 2, alpha=0.15, color="red", label="Underfitting region")
plt.axvspan(3, 6, alpha=0.15, color="green", label="Sweet spot")
plt.axvspan(7, 15, alpha=0.15, color="orange", label="Overfitting region")

plt.xlabel("Tree max_depth (model complexity)")
plt.ylabel("Accuracy")
plt.title("Bias-Variance Tradeoff: Train vs. Test Accuracy")
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Observed Regions (fill from your sweep table)

- Underlifting depth(s): [e.g., 1-2], because train and test accuracy are both low.
- Sweet-spot depth(s): [e.g., 3-6], because test accuracy is highest and the gap is small.
- Overfitting depth(s): [e.g., 8-15], because train accuracy reaches 1.0 while the train-test gap widens.

## Part D: 5-Fold Stratified Cross-Validation

Cross-validation is rin ONLY on the training set.
The held-out test set is never used for models selection.
Mean ± standard deviation shows both performance and stability.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

models = {
    "Shallow tree (depth=1, high bias)": Pipeline([
        ("scaler", StandardScaler()),
        ("model", DecisionTreeClassifier(max_depth=1, random_state=42))
    ]),
    "Moderate tree (depth=4)": Pipeline([
        ("scaler", StandardScaler()),
        ("model", DecisionTreeClassifier(max_depth=4, random_state=42))
    ]),
    "Full tree (no depth limit, high variance)": Pipeline([
        ("scaler", StandardScaler()),
        ("model", DecisionTreeClassifier(max_depth=None, random_state=42))
    ]),
}

cv_results = []

for name, model in models.items():
    scores = cross_val_score(model, X_train, y_train, cv=cv, scoring="accuracy")

    cv_results.append({
        "Model": name,
        "CV mean": scores.mean(),
        "CV std": scores.std(),
        "Fold scores": np.round(scores, 3)
    })

    print(name)
    print(" Fold scores:", np.round(scores, 3))
    print(f" Mean ± std : {scores.mean():.3f} ± {scores.std():.3f}")
    print("-" * 60)

cv_table = pd.DataFrame(cv_results)
print(cv_table.to_string(index=False))
```

### ⚠ Data Leakage Demonstration (DO NOT use in real projects)

Below, scaling is fitted on ALL data (including the test set) BEFORE
cross-validation. This leaks test information into training and usually
produces optimistically biased scores. Compare with the pipeline results above.

```python
leaky_scaler = StandardScaler().fit(X)        # WRONG: uses test data too
X_leaky = leaky_scaler.transform(X)

leaky_model = DecisionTreeClassifier(max_depth=4, random_state=42)

leaky_scores = cross_val_score(leaky_model, X_leaky, y, cv=cv, scoring="accuracy")

print("Leaky CV mean:", leaky_scores.mean())
print("Difference vs. honest pipeline is usually small on this easy dataset,")
print("but on harder datasets leakage inflates scores dangerously.")
```

## Guide Questions and Answers

###1. Which part of this laboratory used supervised learning, and which used unsupervised learning?
**Answer:** Part A employed supervised learning because it relied on explicit target annotations (ground-truth species labels) to train the model to map input features to known categories. In contrast, Part B utilized unsupervised learning, entirely discarding the class labels to let the algorithm autonomously discover latent structures and natural groupings within the feature space.

###2. Did the K-Means clusters match the true species? Explain using the crosstab and Adjusted Rand Index.
**Answer:** Yes, the algorithmic clustering demonstrated a strong alignment with the true biological taxonomy. Iris setosa was isolated into a pure, distinct cluster (50 out of 50 correctly grouped), reflecting its clear linear separability. While versicolor and virginica exhibited some overlapping feature boundaries, the Adjusted Rand Index (ARI) score of approximately 0.73 quantitatively confirms a high degree of agreement between the unsupervised partitions and the actual species classes.

###3. Which max_depth value(s) showed underfitting? What evidence supports this?
**Answer:** A max_depth of 1 (and 2) exhibited underfitting. The definitive evidence is the suppressed performance across both evaluation splits—specifically yielding a low training accuracy and an equally low testing accuracy (hovering around 66.6% for depth 1), indicating that the model is too simple to capture the underlying relationships in the data.

###4. Which max_depth value(s) showed overfitting? What evidence supports this?
**Answer:** max_depth values ranging from 7 to 15 demonstrated clear overfitting. The diagnostic signature is the prominent generalization gap: the training accuracy reaches an asymptotic maximum of 100%, while the testing accuracy plateaus or drops lower at 91.1%, proving the tree has learned training sample quirks rather than generalizable patterns.

###5. Why does training accuracy reach 100% at large depths?
**Answer:** Training accuracy reaches 100% because an unconstrained decision tree recursively partitions the feature space until every leaf node contains samples from a single class. With high tree depth, the model transitions from learning underlying trends to effectively memorizing every individual training instance, including random noise.

###6. What do the CV mean and CV standard deviation tell you about a model?
**Answer:** The CV mean provides an unbiased estimate of the model’s expected out-of-sample predictive performance across unseen data partitions. Meanwhile, the CV standard deviation measures performance stability across folds, revealing how sensitive or robust the model is to small changes in training sample composition.

###7. Why must StandardScaler be placed inside the Pipeline?
**Answer:** Embedding StandardScaler inside a Pipeline is essential to prevent data leakage during cross-validation. If scaling parameters (mean and standard deviation) are calculated prior to splitting, statistics from the validation/testing sets contaminate the training transformations, leading to artificially inflated performance metrics.

###8. Connect your sweep curve to bias and variance.
**Answer:** The hyperparameter sweep curve directly maps out the classic bias-variance tradeoff. Shallow depths represent a regime of high bias (underfitting) due to overly restrictive decision rules. As depth increases, bias decreases, but the model shifts into high variance (overfitting), becoming hypersensitive to training noise; the ideal depth lies at the curve's optimal balance point.
